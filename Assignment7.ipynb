```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv("DataSet7.csv")

df.head()
```

```python
plt.scatter(df['Study_Hours'], df['Final_Score'])

plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.title("Student Score Vs Study Hour")
plt.show()
```

```python
plt.bar(df['Student'], df['English'])

plt.xlabel("Name")
plt.ylabel("English Score")
plt.title("Student Score in English")
plt.show()
```

```python
pieData = df.groupby("Department")['Student'].count()

plt.pie(pieData, labels=pieData.index,  autopct="%1.1f%%")
plt.title("Student Per Deparment")
plt.show()
```
