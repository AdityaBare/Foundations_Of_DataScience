```python
import numpy as np
```

```python
arr = np.array([1,4,7,9,21,12,31,9])
```

```python
range = arr.max()-arr.min()
range
```

```python
mean = arr.mean()
mean
```

```python
arr.var()
```

```python
var =0
for i in arr:
    var= var+((i-mean)**2)

var = var/len(arr)
var
```

```python
arr.std()
```

```python
std = np.sqrt(var)
std
```

# Coefficient of Variation
    * How large is the variation compared with the average

```python
coef = (std*100)/mean
coef
```

```python
first = np.percentile(arr, 25)
second = np.percentile(arr,75)
third = second-first

print("First :", first)
print("Second :", second)
print("Third :", third)
```
