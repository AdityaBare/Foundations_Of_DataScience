```python
# plotting normal distribution graph
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import norm


df = pd.read_csv('StudentsPerformance.csv')

print(df.head(),"\n")
print(df.describe(),"\n")
print(df.info(),"\n")
```

```python
# compute mean and standard deviation
data = df['math score']
mean = np.mean(data)
std_dev = np.std(data)

print("Mean:", mean)
print("Standard Deviation:", std_dev)

```

```python
# generate the normal distribution curve
x = np.linspace(mean - 4*std_dev, mean + 4*std_dev, 1000)
y = norm.pdf(x, mean, std_dev)
```

```python
# plot the normal distribution curve
plt.figure(figsize=(10, 6))
plt.plot(x, y, color='blue', label='Normal Distribution Curve')
plt.hist(data, bins=30, density=True, alpha=0.6, color='orange', label='Math Scores Histogram')
plt.axvline(mean, color='red', linestyle='dashed', linewidth=1, label='Mean')
plt.axvline(mean + std_dev, color='green', linestyle='dashed', linewidth=1, label='Mean + 1 Std Dev')
plt.axvline(mean - std_dev, color='green', linestyle='dashed', linewidth=1, label='Mean - 1 Std Dev')
plt.title('Normal Distribution of Math Scores')
plt.xlabel('Math Scores')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```
