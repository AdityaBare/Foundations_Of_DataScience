```python
import pandas as pd
import numpy as np
from scipy import stats
```

# Manually Caluculating 

```python
age = [12,78,36,28,89,53,28,19,67,87]
```

```python
print("Mean")
total = 0
for i in age:
    total+=i

print(total/len(age))
```

```python
print("Median")
age = sorted(age)

n = len(age)
if n%2 == 1:
    print(age[n//2])
else:
    print((age[n//2]+age[n//2-1])/2)
```

```python
print("Mode")

# `Create freq dictionary not set`
freq = {}

for x in age:
    if x in freq:
        freq[x]+=1
    else:
        freq[x] = 1

max = 0
mode = -1
for x in freq:
    if max< freq[x]:
        max = freq[x]
        mode = x

print(mode)
```

```python
df = pd.DataFrame(age, columns= ['age'])
```

```python
print("Using Pandas")
print("Mean :", df['age'].mean())
print("Median : ", df['age'].median())
print("Mode :",df['age'].mode()[0])
```

```python
print("SCIPY")
print("Mean :",stats.tmean(age))
print("Median : ", np.median(age))
print("Mode : ",stats.mode(age))
```
